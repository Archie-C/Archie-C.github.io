```python
# Base imports you will need for loading and visualising data
# [Remember to install the packages if you want to use them]
# You will also need to install ipykernel if you are runnning this locally
# as a Jupyter notebook.
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
# Other imports you may need - uncomment as you need them
# import numpy as np
# import plotly.express as px
# import sklearn
```

```python
# Load the data from the csv file into a pandas DataFrame
data = pd.read_csv('data_train.csv')
```

```python
# Show the first 5 rows of the DataFrame
data.head()
```

```python
# Describe the DataFrame to get a summary of the data
data.describe()
```

```python
# Scatter plot of one feature against the target variable
# This kind of plot is useful to visualise the relationship between a feature and the target variable
# It may not be very informative if the dataset has many features but it is a good starting point.
# REPLACE 'x_1' with the name of the feature you want to plot against the target variable 'y'.
plt.figure(figsize=(12, 6))
plt.scatter(data['x_1'], data['y'], color='blue', label='Data points', alpha=0.5, s=10)
plt.xlabel('x_1')
plt.ylabel('y')
plt.title('Scatter Plot of x_1 vs y')
plt.show()
```

```python
# Correlation matrix
# This can be useful to identify which features are most correlated, either with each other
# or with the target variable.
# Note: this only works for numerical features.
# Note: Correlation does not imply causation, so be careful when interpreting the results.

plt.figure(figsize=(10, 8))
sns.heatmap(data.corr(), annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Distribution of the target variable
# This can be useful to identify if the target variable is normally distributed or if it is skewed.
plt.figure(figsize=(10, 6))
sns.histplot(data['y'], kde=True, color='blue')
plt.xlabel('y')
plt.ylabel('Frequency')
plt.title('Distribution of y')
plt.show()
```
